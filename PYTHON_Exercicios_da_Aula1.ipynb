# Problema 1 — Classificação da média nas fronteiras

**Nível:** Moderado


Um sistema acadêmico recebe três notas de um aluno.

Crie um programa que:

1. leia as três notas como `float`;
2. calcule a média aritmética;
3. classifique:
   - média menor que `4`: `Reprovado`;
   - média maior ou igual a `4` e menor que `6`: `Recuperação`;
   - média maior ou igual a `6` e menor que `9`: `Aprovado`;
   - média maior ou igual a `9`: `Aprovado com destaque`;
4. exiba a média com duas casas decimais e a classificação.

### Casos obrigatórios de teste

Sua lógica deve classificar corretamente médias exatamente iguais a:

- `4.0`
- `6.0`
- `9.0`

### Microdefesa

Explique em uma linha por que a ordem dos testes pode alterar o resultado.

```python
# ============================================================
# PROBLEMA 1
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Três notas de um alunos.
#
# Processamento: Somar as três notas, dividir por 3 para obter a média aritmética e avaliar a média através das condições (if/elif/else) para determinar a classificação.
#
# Saída: A média do aluno formatada com duas casas decimais e a sua classificação (Reprovado, Recuperação, Aprovado ou Aprovado com destaque).
#

# IMPLEMENTAÇÃO
nota1 = float(input("Digite sua primeira nota: "))
nota2 = float(input("Digite sua segunda nota: "))
nota3 = float(input("Digite sua terceira nota: "))

media = (nota1 + nota2 + nota3) / 3

print(f"Sua média final é {media:.2f} e a classificação é:")

if media >= 9:
    print("Aprovado com destaque")
elif media >= 6:
    print("Aprovado")
elif media >= 4:
    print("Recuperação")
else:
    print("Reprovado")

# MICRODEFESA: por que a ordem das condições pode alterar a classificação?
# Como as condições são lidas do topo para baixo, se um teste mais genérico for feito antes de um mais restrito, o programa pode validar o valor precocemente e atribuir a categoria errada.
```

```python
# ============================================================
# PROBLEMA 5
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Número inteiro maior ou igual a 0.
#
# Processamento: Fazer um laço de repetição para que enquanto n for maior ou igual a 0. Em cada iteração, testar se o número é maior que 0 e divisível por 5 para exibir a mensagem,
# caso contrário, exibir apenas o número.
#
# Saída: A contagem regressiva de n até 0 com a indicação dos números divisíveis por 5.
#

# IMPLEMENTAÇÃO
n = int(input("Digite um número inteiro maior ou igual a 0: "))

while n >= 0:
    if n > 0 and n % 5 == 0:
        print(f"{n} é divisível por 5")


    else:
        print(n)
    n =  n - 1

# MICRODEFESA: qual linha garante que o while possa terminar?
# A linha 'n = n - 1', pois ela reduz o valor de n em cada volta até que ele seja menor que 0, fazendo com que a condição 'n >= 0' do while se torne falsa e o laço termine.
```

# Problema 10 — Função de classificação numérica

**Nível:** Moderado


Crie a função:

`classificar_numero(numero)`

Ela deve retornar exatamente uma das strings:

- `positivo e par`
- `positivo e ímpar`
- `negativo e par`
- `negativo e ímpar`
- `zero`

Depois:

1. leia um inteiro;
2. passe-o como argumento;
3. armazene o retorno;
4. exiba o resultado.

### Casos obrigatórios de teste

- `-4`
- `-3`
- `0`
- `2`
- `5`

### Microdefesa

Explique por que `0` merece uma decisão própria antes de classificá-lo como positivo ou negativo.

```python
# ============================================================
# PROBLEMA 10
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Um número inteiro.
#
# Processamento: Definir a função classificar_numero(numero). Nela, testar se o número é igual a 0.
# Caso contrário, verificar se ele é maior que 0 (positivo) ou menor que 0 (negativo) e, dentro de cada caso, checar a paridade usando o operador resto.
#
# Saída: Indicação se o número é positivo ou negativo, e ímpar ou par.
#

# IMPLEMENTAÇÃO
def classificar_numero(numero):
  if numero == 0:
    return "Zero"

  elif numero > 0:
    if numero % 2 == 0:
      return "Positivo e par"
    else:
      return "Positivo e ímpar"

  else:
    if numero % 2 == 0:
      return "Negativo e par"
    else:
      return "Negativo e ímpar"


valor = int(input("Digite um número inteiro: "))

resultado = classificar_numero(valor)
print(resultado)


# MICRODEFESA: por que zero não deve cair nas categorias positivo/negativo?
# Porque o zero é o elemento neutro da adição, não tem sinal e representa a falta de valor positivo ou negativo.
```

# Problema 11 — Calculadora e retorno None

**Nível:** Moderado/Difícil


Crie:

`calcular(a, b, operacao)`

`operacao` será uma `str` contendo exatamente um destes símbolos:

- `+`
- `-`
- `*`
- `/`

A função deve:

- retornar o resultado da operação válida;
- retornar `None` se a operação não for reconhecida;
- retornar `None` se houver tentativa de divisão por zero.

Depois, o programa deve exibir:

`Operação inválida`

quando o retorno for `None`.

### Casos obrigatórios de teste

- `10, 2, "+"`
- `10, 2, "/"`
- `10, 0, "/"`
- `10, 2, "x"`

### Microdefesa

Explique a diferença entre imprimir o resultado dentro da função e retorná-lo.

```python
# ============================================================
# PROBLEMA 11
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Dois números e uma string com o símbolo da operação desejada.
#
# Processamento: Definir a função calcular(a, b, operacao). Tratar  as condições inválidas: operação desconhecida ou divisão por zero. Para operações válidas, retornar o resultado do cálculo.
# No programa principal, capturar o retorno e exibir "Operação inválida" se o retorno for None, ou o valor numérico caso contrário.
#
# Saída: O resultado da operação matemática ou a mensagem de erro "Operação inválida".
#

# IMPLEMENTAÇÃO
def calcular(a, b, operacao):
  if operacao == "+":
    return a + b
  elif operacao == "-":
    return a - b
  elif operacao == "*" or operacao == "x":
    return a * b
  elif operacao == "/":
    if b == 0:
      return None
    else:
      return a / b
  else:
    return "None"

a = float(input("Digite um número: "))
b = float(input("Digite outro número: "))
operacao = str(input("Operação: "))

resultado = calcular(a, b, operacao)

if resultado is None:
	print("Operação inválida")
else:
  print(resultado)

# MICRODEFESA: por que print dentro da função não substitui return?
# O print apenas mostra um texto na tela do terminal sem guardar a informação. O return envia o valor de volta para o programa,
# permitindo que ele seja armazenado em variáveis, testado em condicionais ou usado em outros cálculos.
```

# Problema 12 — Potência por multiplicações sucessivas

**Nível:** Difícil


Crie:

`potencia(base, expoente)`

A função deve calcular a potência usando multiplicações sucessivas.

Considere `expoente` inteiro maior ou igual a `0`.

### Restrições

- não utilize `**`;
- não utilize bibliotecas.

### Casos obrigatórios de teste

- `potencia(2, 4)` → `16`
- `potencia(5, 0)` → `1`
- `potencia(-2, 3)` → `-8`
- `potencia(-2, 4)` → `16`

### Microdefesa

Explique por que o resultado inicial precisa ser `1`, especialmente quando o expoente é `0`.

```python
# ============================================================
# PROBLEMA 12
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Os parâmetros base e expoente.
#
# Processamento: Definir a função potencia(base, expoente). Inicializar uma variável acumuladora para o resultado com o valor 1. Executar um laço for que repite o número de vezes determinado pelo expoente.
# A cada iteração, multiplicar a variável acumuladora pelo valor da base. Caso o expoente seja 0, o laço não é executado e o valor 1 é mantido.
#
# Saída: O valor numérico da potência calculado e retornado pela função.
#

# IMPLEMENTAÇÃO
def potencia(base, expoente):
  resultado = 1
  for num in range(expoente):
    resultado = resultado * base

  return resultado


base = int(input("Digite um número:"))
expoente = int(input("Digite o expoente maior ou igual a 0: "))

resultado = potencia(base, expoente)
print(resultado)


# MICRODEFESA: por que iniciar resultado em 1 resolve também expoente zero?
# Porque 1 é o elemento neutro da multiplicação, quando o expoente é 0, o laço for faz 0 voltas, fazendo a função retornar o 1 inicial, já que todo número elevado a zero é igual a 1.
```

Explique o papel de `% 10` e de `// 10` no algoritmo.

Explique por que os zeros que aparecem à esquerda do valor invertido não permanecem na representação inteira.

# Problema 15 — Palíndromo preservando o valor original

**Nível:** Difícil


Um inteiro é palíndromo quando possui o mesmo valor quando seus dígitos são lidos nas duas direções.

Crie:

`eh_palindromo(numero)`

A função deve:

1. receber um inteiro positivo;
2. preservar o valor original;
3. construir numericamente o valor invertido;
4. comparar invertido e original;
5. retornar `True` ou `False`.

Depois, exiba:

`Palíndromo`

ou

`Não palíndromo`

### Restrições

- não utilize `str`;
- não utilize coleções;
- não utilize bibliotecas.

### Casos obrigatórios de teste

- `121` → palíndromo
- `1331` → palíndromo
- `1234` → não palíndromo
- `10` → não palíndromo
- `1001` → palíndromo

### Microdefesa

Explique por que modificar diretamente o único valor de `numero` durante a inversão pode impedir a comparação final correta.

```python
# ============================================================
# PROBLEMA 15
# ============================================================

# PREVISÃO E DECOMPOSIÇÃO
# Entrada(s): Um número inteiro positivo.
#
# Processamento: Definir a função eh_palindromo(numero). Salvar o valor inicial do argumento em uma variável auxiliar. Usar um laço while para construir o valor invertido por meio de divisões sucessivas.
# Ao final da inversão, comparar o valor invertido com a variável original. A função deve retornar True se forem iguais ou False se forem diferentes.
#
# Saída: A indicação se o número é "Palíndromo" ou "Não palíndromo".
#

# IMPLEMENTAÇÃO
def eh_palindromo(numero):
  original = numero
  invertido = 0
  while numero > 0:
    digito = numero % 10
    invertido = (invertido * 10) + digito
    numero = numero // 10

  print(invertido)

  if invertido == original:
    return True
  else:
    return False

numero = int(input("Digite um número inteiro positivo"))

resultado = eh_palindromo(numero)

if resultado is True:
	print("Palíndromo")
else:
	print("Não palíndromo")


# MICRODEFESA: qual valor precisa ser preservado e por quê?
# O número original. Ele precisa ser salvo em uma variável auxiliar porque o laço de inversão vai destruindo o número original. Sem a cópia, não teria como comparar o número invertido com o original no final.
```
